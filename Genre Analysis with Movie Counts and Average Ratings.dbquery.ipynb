```sql
SELECT
    TRIM(g.genre) AS Genre,
    COUNT(*) AS `Movie Count`,
    
    ROUND(AVG(md.vote_average), 2) AS `Average Popularity`

FROM movies_dataset md
JOIN (
    SELECT id, SUBSTRING_INDEX(SUBSTRING_INDEX(genre, ',', n.n), ',', -1) AS genre
    FROM movies_dataset
    JOIN (
        SELECT 1 n UNION SELECT 2 UNION SELECT 3 UNION SELECT 4 UNION SELECT 5
    ) n
    ON CHAR_LENGTH(genre) - CHAR_LENGTH(REPLACE(genre, ',', '')) >= n.n - 1
) g ON md.id = g.id
GROUP BY g.genre
ORDER BY  `Average Popularity` DESC LIMIT 20;
```
