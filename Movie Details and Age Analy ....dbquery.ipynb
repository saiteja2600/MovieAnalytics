```sql
SELECT
    md.original_title AS Title,
    md.original_language AS Language,
    md.genre AS Genre,
    md.release_date AS ReleaseDate,
    year(curdate()) - year(md.release_date) as Age
    
FROM movies_dataset md
WHERE md.release_date IS NOT NULL
  AND YEAR(md.release_date) <= YEAR(CURDATE())
LIMIT 10;
```
