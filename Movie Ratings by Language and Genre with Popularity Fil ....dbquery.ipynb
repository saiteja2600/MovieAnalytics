```sql
SELECT
    md.original_language AS language,
    md.genre AS genre,
    ROUND(AVG(md.vote_average), 2) AS Rating
FROM movies_dataset md
WHERE md.original_language IS NOT NULL and md.popularity>10 and md.vote_count>100 and md.vote_average>7
GROUP BY md.original_language,md.genre
ORDER BY MAX(md.popularity) DESC;
```
