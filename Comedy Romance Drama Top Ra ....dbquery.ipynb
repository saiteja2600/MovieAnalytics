```sql
SELECT
    md.original_title AS `Title`,
    md.original_language AS `Language`,
    md.genre AS `Genre`,
    md.popularity AS `Popularity`,
    md.vote_count AS `Vote Count`,
    md.vote_average AS `Average Rating`,
    md.release_date AS `Release Date`
FROM movies_dataset md
WHERE md.genre LIKE '%Comedy%'
  AND md.genre LIKE '%Romance%'
  AND md.genre LIKE '%Drama%'
  AND md.popularity > 10 and md.vote_count > 1000 and md.vote_average > 7
ORDER BY md.vote_average DESC;
```
